# Plugin System Demo

This notebook demonstrates the plugin system built in `plugin_system/`.

**Architecture:**
- `Plugin` – abstract base class every plugin must subclass
- `PluginManager` – registers, activates, and looks up plugins
- `PluginLoader` – discovers and loads plugins from files or directories

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path(".").resolve()))

from plugin_system import PluginManager, PluginLoader

manager = PluginManager()
loader = PluginLoader(manager)

# Load all plugins from the plugins/ directory
count = loader.load_from_directory("plugins")
print(f"Loaded {count} plugin(s)")
```

```python
# List registered plugins
import json
print(json.dumps(manager.list_plugins(), indent=2))
```

```python
# Use the hello plugin
hello = manager.get("hello")
hello.execute("Claude")

# Use the math plugin
math = manager.get("math")
math.execute("add", 7, 3)
math.execute("mul", 4, 5)
```

```python
# Unregister a plugin
manager.unregister("hello")
print("Remaining plugins:", [p["name"] for p in manager.list_plugins()])
```
